```python
import smtplib
from email.message import EmailMessage
import os

# =========================
# 1. Email Credentials
# =========================
sender_email = "gwpmuskan@gmail.com"
receiver_email = "favpython@gmail.com"
app_password = "wjse xald owqc gccr"

# =========================
# 2. Create Email
# =========================
msg = EmailMessage()
msg['Subject'] = 'Excel File from Python'
msg['From'] = sender_email
msg['To'] = receiver_email
msg.set_content('Hello,\n\nPlease find the attached Excel file.\n\nRegards')

# =========================
# 3. Attach Excel File
# =========================
file_path = r"C:\Users\Muska\Desktop\customers.xlsx"

# Check if file exists (important for debugging)
if not os.path.exists(file_path):
    print("❌ File not found! Check the path.")
else:
    with open(file_path, "rb") as f:
        file_data = f.read()
        file_name = os.path.basename(file_path)

    msg.add_attachment(file_data,
                       maintype='application',
                       subtype='octet-stream',
                       filename=file_name)

    # =========================
    # 4. Send Email
    # =========================
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp:
            smtp.login(sender_email, app_password)
            smtp.send_message(msg)
        print("✅ Email sent successfully!")

    except Exception as e:
        print("❌ Error occurred:", e)
```

```python
import schedule
import time
import threading

stop_flag = False   # control variable

def job():
    print("Hello - Background Task")

def run_scheduler():
    while not stop_flag:   # 🔥 infinite nahi
        schedule.run_pending()
        time.sleep(1)

# schedule task
schedule.every(2).seconds.do(job)

# thread start
t = threading.Thread(target=run_scheduler)
t.start()

print("Main program running...")

# main program ka kaam
for i in range(10):
    print("Main work", i)
    time.sleep(1)

# 🔥 stop scheduler
stop_flag = True
t.join()   # thread ko properly band karo

print("Program finished")
```
